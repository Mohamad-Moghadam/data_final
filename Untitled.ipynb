# 📱 Google Play Store Apps — Comprehensive Data Analysis

**Datasets:** `googleplaystore.csv` | `googleplaystore_user_reviews.csv`  
**Goal:** Clean, explore, and visualize trends across 10K+ apps and 64K+ user reviews.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)
plt.rcParams['font.size'] = 12
```

## Load the Datasets

```python
df_apps = pd.read_csv('googleplaystore.csv')
df_reviews = pd.read_csv('googleplaystore_user_reviews.csv')

print("=" * 50)
print("Apps Dataset Shape:", df_apps.shape)
print("Reviews Dataset Shape:", df_reviews.shape)
print("=" * 50)
print("\nApps Columns:")
print(df_apps.columns.tolist())
print("\nReviews Columns:")
print(df_reviews.columns.tolist())
```

## Initial Data Inspection

```python
display(df_apps.head())

print("\n" + "=" * 50)
print("DATAFRAME INFO")
print("=" * 50)
df_apps.info()

print("\n" + "=" * 50)
print("DESCRIPTIVE STATISTICS")
print("=" * 50)
display(df_apps.describe())

print("\n" + "=" * 50)
print("MISSING VALUES PER COLUMN")
print("=" * 50)
missing = df_apps.isnull().sum()
missing_pct = (missing / len(df_apps) * 100).round(2)
missing_df = pd.DataFrame({'Missing Count': missing, 'Missing %': missing_pct})
display(missing_df[missing_df['Missing Count'] > 0])
```

## Data Cleaning & Feature Engineering

```python
df = df_apps.copy()

duplicates = df.duplicated().sum()
df = df.drop_duplicates()

df = df[df['Rating'] <= 5]

df['Reviews'] = pd.to_numeric(df['Reviews'], errors='coerce')

def clean_size(size):
    if isinstance(size, str):
        if 'M' in size:
            return float(size.replace('M', ''))
        elif 'k' in size:
            return float(size.replace('k', '')) / 1024
        elif size == 'Varies with device':
            return np.nan
    return np.nan

df['Size_MB'] = df['Size'].apply(clean_size)

df['Installs'] = df['Installs'].str.replace('[+,]', '', regex=True)
df['Installs'] = pd.to_numeric(df['Installs'], errors='coerce')

df['Price'] = df['Price'].str.replace('$', '', regex=False)
df['Price'] = pd.to_numeric(df['Price'], errors='coerce')

df['Last Updated'] = pd.to_datetime(df['Last Updated'], errors='coerce')
df['Update_Year'] = df['Last Updated'].dt.year

df['Rating'] = df['Rating'].fillna(df['Rating'].median())
df['Reviews'] = df['Reviews'].fillna(0)
df['Size_MB'] = df['Size_MB'].fillna(df['Size_MB'].median())
df['Installs'] = df['Installs'].fillna(df['Installs'].median())
df['Price'] = df['Price'].fillna(0)

df['Log_Reviews'] = np.log1p(df['Reviews'])
df['Log_Installs'] = np.log1p(df['Installs'])
df['Size_Category'] = pd.cut(
    df['Size_MB'], 
    bins=[0, 5, 20, 50, 100, float('inf')], 
    labels=['<5MB', '5-20MB', '20-50MB', '50-100MB', '>100MB']
)

print(f"Duplicates removed: {duplicates}")
print(f"Cleaned Dataset Shape: {df.shape}")
print("\nData types after cleaning:")
print(df[['Rating', 'Reviews', 'Size_MB', 'Installs', 'Price', 'Update_Year']].dtypes)
```

```python
plt.figure(figsize=(12, 7))
sns.histplot(df['Rating'], bins=35, kde=True, color='steelblue', edgecolor='white', linewidth=0.5)
plt.title('Distribution of App Ratings', fontsize=18, fontweight='bold', pad=20)
plt.xlabel('Rating', fontsize=13)
plt.ylabel('Number of Apps', fontsize=13)

plt.axvline(df['Rating'].mean(), color='crimson', linestyle='--', linewidth=2.5,
            label=f'Mean: {df["Rating"].mean():.2f}')
plt.axvline(df['Rating'].median(), color='forestgreen', linestyle='--', linewidth=2.5,
            label=f'Median: {df["Rating"].median():.2f}')
plt.axvline(df['Rating'].mode()[0], color='darkorange', linestyle='--', linewidth=2.5,
            label=f'Mode: {df["Rating"].mode()[0]:.2f}')

plt.legend(fontsize=12, loc='upper left')
plt.xlim(0.5, 5.5)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

print(f"Mean Rating: {df['Rating'].mean():.2f}")
print(f"Median Rating: {df['Rating'].median():.2f}")
print(f"Mode Rating: {df['Rating'].mode()[0]:.2f}")
print(f"Apps with 5.0 Rating: {(df['Rating'] == 5.0).sum():,}")
print(f"Apps with Rating < 3.0: {(df['Rating'] < 3.0).sum():,}")
```

## Chart 8 — Correlation Heatmap (Log-Transformed)

```python
plt.figure(figsize=(12, 10))

numeric_cols = ['Rating', 'Log_Reviews', 'Size_MB', 'Log_Installs', 'Price']
corr_matrix = df[numeric_cols].corr()

mask = np.triu(np.ones_like(corr_matrix, dtype=bool))
sns.heatmap(
    corr_matrix, 
    mask=mask, 
    annot=True, 
    cmap='RdBu_r', 
    vmin=-1, 
    vmax=1,
    square=True, 
    linewidths=1.5, 
    fmt='.2f', 
    cbar_kws={'shrink': 0.8, 'label': 'Correlation'},
    annot_kws={'size': 13, 'fontweight': 'bold'}
)

plt.title('Correlation Heatmap of Numeric Features', fontsize=18, fontweight='bold', pad=20)
labels = ['Rating', 'Log Reviews', 'Size (MB)', 'Log Installs', 'Price']
plt.xticks(range(len(labels)), labels, rotation=45, ha='right')
plt.yticks(range(len(labels)), labels, rotation=0)
plt.tight_layout()
plt.show()
```

## Key Insights & Conclusions

| Insight | Finding |
|---------|---------|
| **Category Volume** | FAMILY and GAME dominate by app count, but COMMUNICATION and SOCIAL lead in total installs |
| **Monetization** | 93%+ of apps are free; paid apps average ~$3.60 with median ~$2.40 |
| **Rating Behavior** | Mean rating is 4.16 with a sharp spike at 5.0, suggesting rating inflation |
| **Size Impact** | No meaningful correlation between app size and user rating |
| **Review Sentiment** | 64% positive, 22% negative, 14% neutral sentiment distribution |
| **Popularity vs Quality** | High review volume does not strongly predict higher ratings |

### Recommendations
1. **For Developers**: Focus on functionality over minimizing app size; ratings are unaffected by storage footprint
2. **For Marketers**: Free apps dominate; consider freemium models over paid upfront pricing
3. **For Analysts**: Use log-transformed Reviews/Installs for correlation analysis due to extreme right skew
