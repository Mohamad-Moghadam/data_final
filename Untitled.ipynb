## Data Cleaning

### Make a copy for cleaning

```python
df = df_apps.copy()
```

### Remove the one row with incorrect data (Rating = 19.0, which is invalid)

```python
df = df[df['Rating'] <= 5]
```

### Clean 'Reviews' column - convert to numeric

```python
df['Reviews'] = pd.to_numeric(df['Reviews'], errors='coerce')
```

### Clean 'Size' column
### Remove 'M' and 'k', convert to MB

```python
def clean_size(size):
    if isinstance(size, str):
        if 'M' in size:
            return float(size.replace('M', ''))
        elif 'k' in size:
            return float(size.replace('k', '')) / 1024
        elif size == 'Varies with device':
            return np.nan
    return np.nan

df['Size_MB'] = df['Size'].apply(clean_size)
```

### Clean 'Installs' column - remove '+' and ',' then convert to numeric

```python
df['Installs'] = df['Installs'].str.replace('[+,]', '', regex=True)
df['Installs'] = pd.to_numeric(df['Installs'], errors='coerce')
```

### Clean 'Price' column - remove '$' and convert to numeric

```python
df['Price'] = df['Price'].str.replace('$', '', regex=False)
df['Price'] = pd.to_numeric(df['Price'], errors='coerce')
```

### Fill missing Ratings with median (optional strategy)

```python
df['Rating'].fillna(df['Rating'].median(), inplace=True)

print("Cleaned Dataset Shape:", df.shape)
print("\nData types after cleaning:")
print(df[['Rating', 'Reviews', 'Size_MB', 'Installs', 'Price']].dtypes)
```

## Chart 1 — Distribution of App Categories

```python
plt.figure(figsize=(14, 8))
category_counts = df['Category'].value_counts().head(15)

sns.barplot(x=category_counts.values, y=category_counts.index, palette='viridis')
plt.title('Top 15 App Categories by Number of Apps', fontsize=16, fontweight='bold')
plt.xlabel('Number of Apps', fontsize=12)
plt.ylabel('Category', fontsize=12)
```

### Add value labels on bars

```python
for i, v in enumerate(category_counts.values):
    plt.text(v + 10, i, str(v), color='black', va='center', fontsize=10)

plt.tight_layout()
plt.show()

print(f"Total unique categories: {df['Category'].nunique()}")
```

## Chart 2 — Distribution of App Ratings

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['Rating'], bins=30, kde=True, color='skyblue', edgecolor='black')
plt.title('Distribution of App Ratings', fontsize=16, fontweight='bold')
plt.xlabel('Rating', fontsize=12)
plt.ylabel('Number of Apps', fontsize=12)
plt.axvline(df['Rating'].mean(), color='red', linestyle='--', 
            label=f'Mean: {df["Rating"].mean():.2f}')
plt.axvline(df['Rating'].median(), color='green', linestyle='--', 
            label=f'Median: {df["Rating"].median():.2f}')
plt.legend()
plt.xlim(1, 5)
plt.tight_layout()
plt.show()

print(f"Mean Rating: {df['Rating'].mean():.2f}")
print(f"Median Rating: {df['Rating'].median():.2f}")
print(f"Apps with 5.0 Rating: {(df['Rating'] == 5.0).sum()}")
```

## Chart 3 — Free vs Paid Apps

```python
plt.figure(figsize=(8, 8))
type_counts = df['Type'].value_counts()

colors = ['#66b3ff', '#ff9999']
plt.pie(type_counts.values, labels=type_counts.index, autopct='%1.1f%%', 
        startangle=90, colors=colors, explode=(0.02, 0.05),
        shadow=True, textprops={'fontsize': 12})

plt.title('Distribution of Free vs Paid Apps', fontsize=16, fontweight='bold')
plt.axis('equal')
plt.tight_layout()
plt.show()

print(f"Free Apps: {type_counts.get('Free', 0)}")
print(f"Paid Apps: {type_counts.get('Paid', 0)}")
```

## Chart 4 — Average Rating by Category (Top 15)

```python
plt.figure(figsize=(14, 8))
avg_rating = df.groupby('Category')['Rating'].mean().sort_values(ascending=False).head(15)

sns.barplot(x=avg_rating.values, y=avg_rating.index, palette='coolwarm')
plt.title('Top 15 Categories by Average Rating', fontsize=16, fontweight='bold')
plt.xlabel('Average Rating', fontsize=12)
plt.ylabel('Category', fontsize=12)
plt.xlim(4.0, 4.5)
```

### Add value labels

```python
for i, v in enumerate(avg_rating.values):
    plt.text(v + 0.01, i, f'{v:.2f}', color='black', va='center', fontsize=10)

plt.tight_layout()
plt.show()
```

## Chart 5 — App Size Distribution

```python
plt.figure(figsize=(10, 6))
```

### Filter out extreme outliers for better visualization

```python
size_data = df['Size_MB'].dropna()
size_data = size_data[size_data <= 100]  # Focus on apps under 100MB

sns.histplot(size_data, bins=40, kde=True, color='coral', edgecolor='black')
plt.title('Distribution of App Sizes (Up to 100 MB)', fontsize=16, fontweight='bold')
plt.xlabel('Size (MB)', fontsize=12)
plt.ylabel('Number of Apps', fontsize=12)
plt.axvline(size_data.mean(), color='red', linestyle='--', 
            label=f'Mean: {size_data.mean():.1f} MB')
plt.axvline(size_data.median(), color='green', linestyle='--', 
            label=f'Median: {size_data.median():.1f} MB')
plt.legend()
plt.tight_layout()
plt.show()

print(f"Mean Size: {size_data.mean():.2f} MB")
print(f"Median Size: {size_data.median():.2f} MB")
```

## Chart 6 — Correlation Heatmap of Numeric Features

## Chart 7 — Sentiment Analysis from User Reviews

### Clean reviews data - drop rows with missing sentiment

```python
df_rev_clean = df_reviews.dropna(subset=['Sentiment'])

plt.figure(figsize=(8, 6))
sentiment_counts = df_rev_clean['Sentiment'].value_counts()

colors = ['#2ecc71', '#e74c3c', '#f39c12']  # Green, Red, Orange
sns.barplot(x=sentiment_counts.index, y=sentiment_counts.values, palette=colors)

plt.title('Distribution of User Review Sentiments', fontsize=16, fontweight='bold')
plt.xlabel('Sentiment', fontsize=12)
plt.ylabel('Number of Reviews', fontsize=12)
```

### Add value labels on top of bars

```python
for i, v in enumerate(sentiment_counts.values):
    plt.text(i, v + 50, str(v), ha='center', fontsize=11, fontweight='bold')

plt.tight_layout()
plt.show()

print("Sentiment Distribution:")
print(sentiment_counts)
print(f"\nPercentages:")
print((sentiment_counts / sentiment_counts.sum() * 100).round(2))
```
